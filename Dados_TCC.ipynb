```python
import pandas as pd
```

```python
dados = pd.read_csv('/content/scopus (3).csv')
dados_filtrados = dados[dados['Cited by'] > 20]
dados_filtrados
```

```python
citacoes_ordenada = dados_filtrados.groupby('Title')['Cited by'].mean()
citacoes_ordenada = citacoes_ordenada.sort_values(ascending=False)
```

```python
df_resultado = citacoes_ordenada.reset_index()
```

```python
df_resultado.to_csv('/content/citacoes_ordenada.csv', index = False)
```

```python
df_resultado.to_excel('/content/citacoes_ordenada.xlsx', index=False)
```
